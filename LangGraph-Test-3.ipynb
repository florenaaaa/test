```python
from typing import List, TypedDict
from langgraph.graph import StateGraph
```

```python
class AgentState(TypedDict):
    name: str
    age: str
    skills: List[str]
    final: str
```

```python
def first_node(state: AgentState) -> AgentState:
    """This is the first node"""

    state["final"] = f"{state["name"]}, welcome to the system! "

    return state

def second_node(state: AgentState) -> AgentState:
    """This is the second node"""

    state["final"] = state["final"] + f"You are {state["age"]} years old! "

    return state

def third_node(state: AgentState) -> AgentState:
    """This is the third node"""

    skills_str = ""
    for index, element in enumerate(skills):
        if(index != (len(skills)-1)):
            skills_str = skills_str + element + ","
        else:
            skills_str += "and " + element

    state["final"] = state["final"] + f"You have skills in: " + skills_str

    return state
```

```python
graph = StateGraph(AgentState)

graph.add_node("first_node", first_node)
graph.add_node("second_node", second_node)
graph.add_node("third_node", third_node)

graph.set_entry_point("first_node")
graph.add_edge("first_node", "second_node")
graph.add_edge("second_node", "third_node")
graph.set_finish_point("third_node")

app = graph.compile()
```

```python
result = app.invoke({"name": "Linda", "age": 31, "skills": ["Python", "Machine Learning", "LangGraph"]})
print(result["final"])
```
